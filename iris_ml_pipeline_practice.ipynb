# ML pipeline: Iris flower classifier

## 1. Load the data

The built-in Iris dataset contains **150 flowers**, each with **four measurements** and a species label.

- `X` = features (the measurements used to predict)
- `y` = target (the correct species we want to predict)

Using a built-in dataset removes the need to download external data.

```python
from sklearn.datasets import load_iris
from IPython.display import display

iris = load_iris(as_frame=True)
X = iris.data
y = iris.target

print("Number of flowers:", len(X))
print("Measurements:", list(X.columns))
print("Species:", list(iris.target_names))
display(X.head())
```

## 2. Split into training and testing data

We train on **80% of the flowers** and hold out **20%** to check how well the trained model predicts on data it hasn't seen. `random_state=42` lets someone rerun the same split. `stratify=y` maintains species proportions in both sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training flowers:", len(X_train))
print("Testing flowers:", len(X_test))
```

## 3. Create and train a real scikit-learn Pipeline

The pipeline has two connected stages:

**StandardScaler:** adjusts feature scales using only the training set.

**LogisticRegression:** learns to identify the species from the measurements. Despite its name, logistic regression is a *classification* algorithm.

`.fit()` trains both stages in the correct order. Keeping the scaler inside the pipeline helps avoid accidentally using test data during training.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)

model.fit(X_train, y_train)
print("Model trained successfully!")
```

## 4. Predict and evaluate

`.predict(X_test)` asks the trained model for its guesses about the unseen 20%.

**Accuracy** = correct guesses ÷ total guesses. A score of 0.97 means about 97% of test cases were classified correctly. The notebook prints your actual result.

```python
from sklearn.metrics import accuracy_score, classification_report

predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print(f"TEST ACCURACY: {accuracy:.1%}")
print("\nClassification report:")
print(classification_report(y_test, predictions, target_names=iris.target_names))
```

## 5. Inspect some predictions

This helps you see what the model actually did. `actual` is the correct species, while `predicted` is the model's guess.

```python
import pandas as pd

preview = pd.DataFrame({
    "actual": iris.target_names[y_test.to_numpy()[:10]],
    "predicted": iris.target_names[predictions[:10]],
})
display(preview)
```
